```python
for split in ["train", "valid", "test"]:

    images, labels, missing_labels, missing_images = check_image_label_matching(
        f"{split}/images",
        f"{split}/labels"
    )

    print(f"\n===== {split.upper()} =====")

    print("Images:", len(images))
    print("Labels:", len(labels))

    print("Images không có label:", len(missing_labels))
    print("Labels không có image:", len(missing_images))

    if missing_labels:
        print("Sample missing labels:", list(missing_labels)[:5])

    if missing_images:
        print("Sample missing images:", list(missing_images)[:5])
```

Đọc data.yaml

```python
with open("data.yaml", "r", encoding="utf-8") as file:
    data = yaml.safe_load(file)

print(data)
```

Hiển thị danh sách class

```python
print("Classes:")

for i, name in enumerate(data["names"]):
    print(f"{i}: {name}")
```

Kiểm tra nc và số class

```python
num_classes = data["nc"]
class_names = data["names"]

print("nc =", num_classes)
print("Number of names =", len(class_names))

if num_classes == len(class_names):
    print("OK: nc và số lượng class khớp nhau.")
else:
    print("WARNING: nc và số lượng class không khớp.")
```
