```python
from ultralytics import YOLO
from pathlib import Path
import yaml
import os
```

Khai báo đường dẫn dataset


```python
PROJECT_DIR = Path("..")

DATASET_DIR = PROJECT_DIR / "Obstacle-Detection-Dataset-YOLO"

DATA_YAML = DATASET_DIR / "data.yaml"

print("Dataset:")
print(DATASET_DIR.resolve())

print("\ndata.yaml:")
print(DATA_YAML.resolve())
```

Kiểm tra dataset


```python
if not DATASET_DIR.exists():
    raise FileNotFoundError(
        "Không tìm thấy dataset. "
        "Hãy chạy 01_download_dataset.ipynb trước."
    )

if not DATA_YAML.exists():
    raise FileNotFoundError(
        "Không tìm thấy data.yaml."
    )

print("Dataset và data.yaml tồn tại.")
```

Load YOLOv8


```python
MODEL_NAME = "yolov8n.pt"

model = YOLO(MODEL_NAME)

print("Model loaded:", MODEL_NAME)
```

Thiết lập tham số training


```python
EPOCHS = 50
IMAGE_SIZE = 640
BATCH_SIZE = 8

PROJECT_NAME = "yolo_runs"
RUN_NAME = "road_obstacle"

print("Training configuration:")
print("Epochs     :", EPOCHS)
print("Image size :", IMAGE_SIZE)
print("Batch size :", BATCH_SIZE)
print("Project    :", PROJECT_NAME)
print("Run name   :", RUN_NAME)
```

Train YOLOv8


```python
results = model.train(
    data=str(DATA_YAML),
    epochs=EPOCHS,
    imgsz=IMAGE_SIZE,
    batch=BATCH_SIZE,
    project=PROJECT_NAME,
    name=RUN_NAME,
    save=True,
    save_period=1,
    exist_ok=True
)
```
